```python
class NoiseScheduler:
    def __init__(self, T=1000, schedule="linear", beta_start=0.0001, beta_end=0.02):
        self.T = T
        if schedule == "linear":
            self.betas = torch.linspace(beta_start, beta_end, T)
            self.alphas = 1 - self.betas
            self.alpha_bars = torch.cumprod(self.alphas, dim=0)
        elif schedule == "cosine":
            s = 0.008
            t = torch.arange(0, T+1)
            alpha_bars = torch.cos(((t/T + s) / (1 + s)) * torch.pi / 2) ** 2
            alpha_bars = alpha_bars / alpha_bars[0]
            self.betas = (1 - alpha_bars[1:] / alpha_bars[:-1]).clamp(0, 0.999)
            self.alphas = 1 - self.betas
            self.alpha_bars = torch.cumprod(self.alphas, dim=0)
        
    def add_noise(self, x0, t, epsilon):
        # t is an integer or batch of integers
        device = x0.device
        alpha_bar_t = self.alpha_bars.to(t.device)[t]
        
        # reshape for broadcasting with image dimensions (B, C, H, W)
        while alpha_bar_t.dim() < x0.dim():
            alpha_bar_t = alpha_bar_t.unsqueeze(-1)
        
        xt = torch.sqrt(alpha_bar_t) * x0 + torch.sqrt(1 - alpha_bar_t) * epsilon
        return xt
```

```python
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))  # normalize to [-1, 1]
])

dataset = datasets.CIFAR10(root='../data', train=True, download=True, transform=transform)
dl = DataLoader(dataset, batch_size=32, shuffle=True)
```

```python
# 2. Extract a single batch
for xb, yb in dl:
    # Get the first image in the batch and rearrange to [H, W, C]
    raw_normalized_img = xb[0].permute(1, 2, 0)
    
    # Create the denormalized version (reversing: x * 0.5 + 0.5)
    properly_denormalized_img = (raw_normalized_img * 0.5) + 0.5
    
    # 3. Plot the comparison side-by-side
    fig, axes = plt.subplots(1, 2, figsize=(8, 4))
    
    # Left Plot: The raw normalized image (Will trigger a matplotlib warning)
    axes[0].imshow(raw_normalized_img.numpy())
    axes[0].set_title("Raw Normalized Image\n(Passed Directly to imshow)")
    axes[0].axis('off')
    
    # Right Plot: The denormalized image (What the human eye should see)
    axes[1].imshow(properly_denormalized_img.numpy())
    axes[1].set_title("Properly Denormalized Image\n(Matches Original)")
    axes[1].axis('off')
    
    plt.tight_layout()
    plt.show()
    
    break
```

```python
noise_scheduler = NoiseScheduler()

# Get one batch, take first image
xb, _ = next(iter(dl))
x0 = xb[0:1]  # shape (1, 1, 28, 28), just one image

epsilon = torch.randn_like(x0)
timesteps = [0, 250, 500, 750, 999]

fig, axes = plt.subplots(1, 5, figsize=(12, 3))

for ax, t in zip(axes, timesteps):
    xt = noise_scheduler.add_noise(x0=x0, t=t, epsilon=epsilon)
    img = xt[0, 0].detach().cpu().numpy()  # (28, 28)
    ax.imshow(img, cmap='gray')
    ax.set_title(f't={t}')
    ax.axis('off')

plt.suptitle('Forward Diffusion Process')
plt.tight_layout()
plt.savefig('forward_diffusion.png', dpi=150)
plt.show()
```

```python
linear_ns = NoiseScheduler(schedule='linear')
cosine_ns = NoiseScheduler(schedule='cosine')

plt.figure(figsize=(10, 4))
plt.plot(linear_ns.betas, label='Linear')
plt.plot(cosine_ns.betas, label='Cosine')
plt.legend()
plt.title('Beta Schedules')
plt.show()
```
